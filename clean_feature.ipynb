## Note on data handling

UVL files are indentation- and newline-sensitive: the feature hierarchy 
is represented through indentation, so small formatting issues can prevent 
a model from parsing. During this work, one model required reformatting 
before it would load correctly. The multi-model pipeline below handles 
such cases gracefully using try/except, so a single malformed model does 
not stop the analysis.

## 1. We loaded the feature model

```python
from flamapy.interfaces.python.flamapy_feature_model import FLAMAFeatureModel

fm = FLAMAFeatureModel(
    "/Users/abishekshrestha/Documents/Flamapy/smartwatch_clean.uvl"
)
```

## 2. We checked whether the model is valid

## 3. We found the core features

## 4. We checked for dead features

## 5. We checked false-optional features

```python
fm.false_optional_features()
```

## 6. We generated valid configurations

```python
samples = fm.sampling(size=5)
```

## 8. We converted the results into Pandas

```python
if len(samples) > 0:
    # Get the column headers from the keys
    headers = samples[0].keys()
    
    # Print the header row
    header_line = "".join(f"{key:<15}" for key in headers)
    print(header_line)
    print("-" * len(header_line))
    
    # Print each row of data
    for sample in samples:
        row_line = "".join(f"{str(sample[key]):<15}" for key in headers)
        print(row_line)
```

```python
from flamapy.interfaces.python.flamapy_feature_model import FLAMAFeatureModel
import pandas as pd

# Each model: (friendly name, file path)
models = [
    ("SmartWatch",       "/Users/abishekshrestha/Documents/Flamapy/smartwatch_clean.uvl"),
    ("Xiaomi SmartBand", "/Users/abishekshrestha/Documents/Flamapy/models1.uvl"),
    ("Data VIZ",         "/Users/abishekshrestha/Documents/Flamapy/models2.uvl"),
    ("Pizza",            "/Users/abishekshrestha/Documents/Flamapy/models3.uvl"),
]

results = []

for name, model_path in models:
    try:
        fm = FLAMAFeatureModel(model_path)
        row = {
            "model": name,
            "satisfiable": fm.satisfiable(),
            "num_configurations": len(fm.configurations()),
            "num_core_features": len(fm.core_features()),
            "num_dead_features": len(fm.dead_features()),
            "num_false_optional_features": len(fm.false_optional_features()),
        }
        results.append(row)
        print(f"✓ Processed: {name}")
    except Exception as e:
        print(f"✗ Failed on {name}: {e}")

df = pd.DataFrame(results)

# Save the labelled dataset
df.to_csv("/Users/abishekshrestha/Documents/Flamapy/feature_model_analysis.csv", index=False)
print("Saved results to feature_model_analysis.csv")

df
```
